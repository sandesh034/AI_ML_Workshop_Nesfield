#### **Variables in Python**

A **variable** in Python is a named reference to a memory location where data is stored. Variables make it easy to manipulate and access data throughout a program.


#### **Variable Declaration and Assignment**

Variables are created when you assign a value to them. Python doesn’t require explicit variable declarations or type specifications. In Python, variables are dynamically typed, meaning you don’t need to declare the type explicitly. Python determines the type based on the assigned value.

#### Syntax:
```python
variable_name = value
```


#### **Rules for Naming Variables**
- Must start with a letter or an underscore (`_`).
- Can only contain letters, numbers, and underscores.
- Case-sensitive
- Cannot use reserved keywords 

````python
x = 10
name = "Alice"
is_active = True
````


#### **Multiple Assignments**

````python
a, b, c = 1, 2, 3
print(a, b, c)  # Output: 1 2 3

x = y = z = 0  # All variables assigned to 0
````


#### **Constants**

Python does not have built-in constants, but by convention, variables written in **ALL CAPS** are treated as constants.


````python
PI = 3.14159
GRAVITY = 9.8
````


### 1. **Arithmetic Operators**
| Operator | Description            | Example         |
|----------|------------------------|-----------------|
| `+`      | Addition               | `3 + 2 = 5`     |
| `-`      | Subtraction            | `3 - 2 = 1`     |
| `*`      | Multiplication         | `3 * 2 = 6`     |
| `/`      | Division               | `3 / 2 = 1.5`   |
| `%`      | Modulus                | `3 % 2 = 1`     |
| `**`     | Exponentiation         | `3 ** 2 = 9`    |
| `//`     | Floor division         | `3 // 2 = 1`    |



### 2. **Comparison Operators**
| Operator | Description                 | Example          |
|----------|-----------------------------|------------------|
| `==`     | Equal to                    | `3 == 2 → False` |
| `!=`     | Not equal to                | `3 != 2 → True`  |
| `>`      | Greater than                | `3 > 2 → True`   |
| `<`      | Less than                   | `3 < 2 → False`  |
| `>=`     | Greater than or equal to    | `3 >= 2 → True`  |
| `<=`     | Less than or equal to       | `3 <= 2 → False` |



### 3. **Logical Operators**
| Operator | Description       | Example                     |
|----------|-------------------|-----------------------------|
| `and`    | Logical AND       | `True and False → False`    |
| `or`     | Logical OR        | `True or False → True`      |
| `not`    | Logical NOT       | `not True → False`          |



### 4. **Bitwise Operators**
| Operator | Description              | Example               |
|----------|--------------------------|-----------------------|
| `&`      | Bitwise AND              | `0b101 & 0b011 = 0b001` |
| `|`      | Bitwise OR               | `0b101 | 0b011 = 0b111` |
| `^`      | Bitwise XOR              | `0b101 ^ 0b011 = 0b110` |
| `~`      | Bitwise NOT              | `~0b101 = -0b110`     |
| `<<`     | Left shift               | `0b101 << 2 = 0b10100` |
| `>>`     | Right shift              | `0b101 >> 2 = 0b1`    |



### 5. **Assignment Operators**
| Operator | Description          | Example             |
|----------|----------------------|---------------------|
| `=`      | Assignment           | `x = 5`            |
| `+=`     | Add and assign        | `x += 3 → x = x + 3`|
| `-=`     | Subtract and assign   | `x -= 3 → x = x - 3`|
| `*=`     | Multiply and assign   | `x *= 3 → x = x * 3`|
| `/=`     | Divide and assign     | `x /= 3 → x = x / 3`|
| `%=`     | Modulus and assign    | `x %= 3 → x = x % 3`|
| `//=`    | Floor div and assign  | `x //= 3 → x = x // 3`|
| `**=`    | Power and assign      | `x **= 3 → x = x ** 3`|
| `&=`     | Bitwise AND and assign| `x &= 3 → x = x & 3`|
| `|=`     | Bitwise OR and assign | `x |= 3 → x = x | 3`|
| `^=`     | Bitwise XOR and assign| `x ^= 3 → x = x ^ 3`|
| `<<=`    | Left shift and assign | `x <<= 3 → x = x << 3`|
| `>>=`    | Right shift and assign| `x >>= 3 → x = x >> 3`|


### 6. **Identity Operators**
| Operator | Description              | Example                 |
|----------|--------------------------|-------------------------|
| `is`     | Returns True if objects are identical | `x is y`          |
| `is not` | Returns True if objects are not identical | `x is not y` |



### 7. **Membership Operators**
| Operator | Description                    | Example                  |
|----------|--------------------------------|--------------------------|
| `in`     | Returns True if a value is in a sequence | `'a' in 'abc' → True`|
| `not in` | Returns True if a value is not in a sequence | `'x' not in 'abc' → True`|
