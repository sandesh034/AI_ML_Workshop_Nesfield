## Part 1: Introduction to File Handling

### Why do we need File Handling?

- Programs need to **persist data** beyond runtime
- Read configurations, datasets, and logs
- Write results, reports, and processed data
- **Essential for data science and ML pipelines** — loading datasets, saving models, logging results

### Common File Types:
| Type | Extension | Use Case |
|------|-----------|----------|
| Text | .txt | Simple data, logs |
| CSV | .csv | Tabular data, datasets |
| JSON | .json | Configuration, APIs |
| Binary | .pkl, .h5 | ML models, images |

### Opening and Closing Files

The `open()` function is used to open files. Always remember to close files after use!

```python
# Method 1: Basic file opening (NOT recommended)
file = open("data.txt", "w")
file.write("bmchghdghjcgahjdghjgdhjs")
file.close()  # Must manually close!

```

```python
# Method 2: Using 'with' statement (RECOMMENDED)
# File is automatically closed when the block ends
with open("data.txt", "r") as file:
    content = file.read()
    print(content)
```

### File Open Modes

| Mode | Description |
|------|-------------|
| `"r"` | Read (default) — file must exist |
| `"w"` | Write — creates new or **overwrites** existing |
| `"a"` | Append — adds to end of file |
| `"x"` | Exclusive create — fails if file exists |
| `"r+"` | Read and write |
| `"b"` | Binary mode (e.g., `"rb"`, `"wb"`) |

> ⚠️ **Warning:** Using `"w"` mode will erase all existing content!

## Part 2: Reading Text Files 

```python
# First, let's create a sample text file to work with
with open("poem.txt", "w") as f:
    f.write("Python is awesome,\n")
    f.write("And so are you!\n")
    f.write("Line 5 of the poem.\n")

print("poem.txt created successfully!")
```

```python
# Method 1: read() - Read entire file as one string
with open("poem.txt", "r") as f:
    content = f.read()
    print(content)
```

```python
# Method 3: readlines() - Read all lines into a list
with open("poem.txt", "r") as f:
    lines = f.readlines()
    # print(lines)
    print(f"Number of lines: {len(lines)}")
    print(f"First line: {lines[0].strip()}")
    print(f"Last line: {lines[-1].strip()}")
```

```python
# Method 4: Iterating line by line (MOST MEMORY EFFICIENT for large files)
with open("poem.txt", "r") as f:
    for i, line in enumerate(f, 1):
        print(f"Line {i}: {line.strip()}")
```

## Part 3: Working with CSV Files

### What is CSV?
- **CSV** = Comma-Separated Values
- A simple, widely-used format for storing **tabular data**
- Each line represents a **row**; columns are separated by commas
- First row is usually the **header** (column names)

Python provides a built-in `csv` module for reading and writing CSV files.

```python
import csv

# First, let's create a sample CSV file
header = ["name", "age", "grade"]
students = [
    ["Alice", 20, "A"],
    ["Bob", 22, "B"],
    ["Charlie", 21, "A"],
    ["Diana", 23, "B+"],
    ["Eve", 20, "A-"]
]

with open("students.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(header)       # Write header
    writer.writerows(students)    # Write all data rows

print("students.csv created successfully!")

print("\nFile content:")

with open("students.csv", "r") as f:
    print(f.read())
```

```python
# Reading CSV using csv.reader()
import csv

with open("students.csv", "r") as f:
    reader = csv.reader(f)
    header = next(reader)  # Get the header row
    print(f"Columns: {header}")
    print("-" * 30)
    for row in reader:
        print(f"Name: {row[0]}, Age: {row[1]}, Grade: {row[2]}")
```

## Part 5: Exception Handling

### What Are Exceptions?
- Errors that occur **during program execution** (runtime errors)
- Without handling, they **crash** the program
- Python provides `try/except` to handle them gracefully

### Common Exceptions:
| Exception | When it occurs |
|-----------|---------------|
| `FileNotFoundError` | File doesn't exist |
| `ValueError` | Wrong value type (e.g., `int("hello")`) |
| `PermissionError` | No access rights |
| `ZeroDivisionError` | Division by zero |
| `TypeError` | Wrong type operation |

```python
#Basic try/except
try:
    num = int("hello")
except ValueError:
    print("Error: Cannot convert 'hello' to integer!")
# num = int("hello")
print("Program continues running after the error is handled!")
```

```python
# Multiple except blocks - handling different errors
try:
    file = open("datas.txt", "r")
    content = file.read()
    number = int(content)
    print(f"The number is: {number}")
except FileNotFoundError:
    print("Error: File does not exist!")
except ValueError:
    print("Error: File doesn't contain a valid number!")
except Exception as e:
    print(f"Unexpected error: {e}")
finally:
    print("This block ALWAYS runs (cleanup code goes here).")
```

```python
# Safe File Reading Pattern - combining 'with' and try/except
def read_file_safe(filename):
    """Safely read a file with proper error handling."""
    try:
        with open(filename, "r") as f:
            return f.read()
    except FileNotFoundError:
        print(f"Error: '{filename}' not found.")
        return None
    except PermissionError:
        print(f"Error: No permission to read '{filename}'.")
        return None

# Test with existing file
content = read_file_safe("poem.txt")
if content:
    print("File read successfully:")
    print(content)

# Test with non-existing file
content = read_file_safe("missing_file.txt")
```

## Part 6: Modules

### What is a Module?
- A **module** is simply a `.py` file containing Python code
- Modules help **organize** code into reusable, logical units
- Python has thousands of **built-in** and **third-party** modules

### Benefits:
1. **Code Reusability** — Write once, use everywhere
2. **Organization** — Break large programs into manageable files
3. **Namespace Management** — Avoid naming conflicts

### Popular ML Modules:
`numpy`, `pandas`, `scikit-learn`, `matplotlib`, `tensorflow`

```python
# Different ways to import modules

# 1. Import entire module
import math
print(f"Square root of 16: {math.sqrt(16)}")
print(f"Value of pi: {math.pi}")

# 2. Import specific functions
from math import sqrt, pi, ceil, floor
print(f"\nsqrt(25) = {sqrt(25)}")
print(f"ceil(4.3) = {ceil(4.3)}")
print(f"floor(4.7) = {floor(4.7)}")

# 3. Import with alias
import math as m
print(f"\nUsing alias: m.pow(2, 10) = {m.pow(2, 10)}")

# 4. Import everything (NOT recommended - pollutes namespace)
# from math import *
```

### Creating Your Own Module

Let's create a custom module called `myutils.py`:

```python
# Step 1: Create a module file (myutils.py)
module_code = '''# myutils.py - Our custom module

def greet(name):
    """Return a greeting message."""
    return f"Hello, {name}!"

def add(a, b):
    """Add two numbers."""
    return a + b

def multiply(a, b):
    """Multiply two numbers."""
    return a * b

def is_even(n):
    """Check if a number is even."""
    return n % 2 == 0

PI = 3.14159
COURSE_NAME = "AI and Machine Learning"
'''

with open("myutils.py", "w") as f:
    f.write(module_code)

print("myutils.py module created!")
```

```python
# Step 2: Import and use our custom module
import myutils

print(myutils.greet("Alice"))
print(myutils.greet("Bob"))
print(f"3 + 5 = {myutils.add(3, 5)}")
print(f"4 * 7 = {myutils.multiply(4, 7)}")
print(f"Is 10 even? {myutils.is_even(10)}")
print(f"Is 7 even? {myutils.is_even(7)}")
print(f"PI = {myutils.PI}")
print(f"Course: {myutils.COURSE_NAME}")
```
