## The 6 Core Concepts of OOP

---

## 1. Class

A **class** is a **blueprint** or **template** for creating objects.

- Defines the **attributes** (data) and **methods** (behavior) that objects will have
- Does **not** occupy memory for data until an object is created
- Think of it as an **architectural plan** — it describes the structure but is not the building itself

### Key Points:
- Created using the `class` keyword
- Contains `__init__()` for initialization
- Can have multiple methods and attributes

```python
class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

    def drive(self):
        return f"{self.brand} is driving!"

# The class is defined but NO object exists yet
print(type(Car))
```

## 2. Object

An **object** is a specific **instance** of a class.

- Each object has its own **unique data** but shares the class's methods
- You can create **multiple objects** from the same class
- Objects are the **real entities** that exist in memory

### Key Points:
- Created by calling the class like a function: `ClassName()`
- Each object is independent — changing one doesn't affect another
- `Class` = Blueprint → `Object` = Real thing built from that blueprint

```python
# Creating objects (instances) from the Car class
car1 = Car("Toyota", "Red")
car2 = Car("Honda", "Blue")

print(car1.brand)    # Toyota
print(car2.brand)    # Honda
print(car1.drive())  # Toyota is driving!

# Objects are independent 
print(f"\ncar1 is car2? {car1 is car2}")  # False
print(f"Type of car1: {type(car1)}")      # <class '__main__.Car'>
```

## 3. Inheritance

A child class **inherits** attributes and methods from a parent class.

- Promotes **code reuse** — no need to rewrite common functionality
- Child class can **add new** features or **override** parent behavior
- Use `super()` to call the parent class's methods

### Types of Inheritance:
| Type | Description |
|------|-------------|
| **Single** | One child inherits from one parent |
| **Multiple** | One child inherits from multiple parents |
| **Multilevel** | A → B → C (chain of inheritance) |
| **Hierarchical** | Multiple children inherit from one parent |

```python
# INHERITANCE — Child class inherits from parent class

class Animal:          # Parent
    def __init__(self, name):
        self.name = name
    def speak(self):
        return "Some sound"

class Dog(Animal):     # Child
    def speak(self):
        return f"{self.name}: Woof!"

dog = Dog("Buddy")
print(dog.speak())    # Buddy: Woof!
print(dog.name)       # Buddy (inherited)
```

## 4. Abstraction

**Hiding complex implementation details** and showing only the **essential features** to the user.

- Users interact with a **simple interface** without knowing internal workings
- Achieved using **abstract classes** (from `abc` module)
- Abstract classes **cannot be instantiated** directly
- Forces child classes to **implement required methods**

### Real-World Analogy:
- **TV Remote**: You press buttons (interface) without knowing the electronic circuit inside
- **ATM Machine**: You use simple buttons to withdraw money without knowing the banking system behind it

```python
# ABSTRACTION — Hide complexity, show only essentials

from abc import ABC, abstractmethod

class Shape(ABC):    # Abstract
    @abstractmethod
    def area(self):  # Must implement
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius ** 2

c = Circle(5)
print(c.area())  # 78.5

# Cannot instantiate abstract class directly
try:
    s = Shape()
except TypeError as e:
    print(f"Error: {e}")  # s = Shape() — ERROR!
```

## 5. Polymorphism

**"Many forms"** — the same method name behaves **differently** depending on the object type.

- Allows writing **generic code** that works with objects of different classes
- Achieved through **method overriding** and **duck typing** in Python
- You don't need to know the exact type — just call the method!


```python
# POLYMORPHISM — Same method name, different behavior

class Cat:
    def speak(self):
        return "Meow!"

class Dog:
    def speak(self):
        return "Woof!"

class Duck:
    def speak(self):
        return "Quack!"

# Same name, different behavior
cat=Cat()
dog=Dog()
duck=Duck()

print(cat.speak())
print(dog.speak())
print(duck.speak())
```

## 6. Encapsulation

**Bundling data and methods** together inside a class and **restricting direct access** to internal data.

- Uses **private attributes** (prefix with `__`) to hide data
- Access is controlled through **getter/setter** methods
- Prevents accidental modification of critical data

### Access Modifiers in Python:
| Syntax | Access Level | Example |
|--------|-------------|---------|
| `name` | **Public** — accessible everywhere | `self.name` |
| `_name` | **Protected** — convention only | `self._name` |
| `__name` | **Private** — name mangling applied | `self.__name` |

```python
# ENCAPSULATION — Bundle data + restrict access

class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance

acc = BankAccount(1000)
acc.deposit(500)
print(acc.get_balance())  # 1500

# Trying to access private attribute directly (FAILS)
try:
    print(acc.__balance)
except AttributeError as e:
    print(f"Error: {e}")  # acc.__balance — ERROR!
```

## Part 7: Practice — Bank Account Class

Let's build a complete **Bank Account** system using everything we've learned:
- Encapsulation (private balance)
- Methods with validation
- Inheritance (SavingsAccount)
- Transaction history

```python
# Bank Account Class
class BankAccount:
    """A bank account with deposit, withdraw, and transaction history."""
    
    bank_name = "Nepal Bank Ltd"  # Class attribute
    
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance  # Private attribute
        self.transactions = []    # Transaction history
        print(f"Account created for {owner} with balance Rs.{balance}")
    
    def deposit(self, amount):
        """Deposit money into the account."""
        if amount > 0:
            self.__balance += amount
            self.transactions.append(f"Deposit: +Rs.{amount}")
            print(f"Rs.{amount} deposited. New balance: Rs.{self.__balance}")
        else:
            print("Error: Deposit amount must be positive!")
    
    def withdraw(self, amount):
        """Withdraw money from the account."""
        if amount <= 0:
            print("Error: Withdrawal amount must be positive!")
        elif amount > self.__balance:
            print(f"Insufficient funds! Available: Rs.{self.__balance}")
        else:
            self.__balance -= amount
            self.transactions.append(f"Withdraw: -Rs.{amount}")
            print(f"Rs.{amount} withdrawn. New balance: Rs.{self.__balance}")
    
    def get_balance(self):
        """Return current balance."""
        return self.__balance
    
    def show_transactions(self):
        """Display transaction history."""
        print(f"\n--- Transaction History for {self.owner} ---")
        if not self.transactions:
            print("No transactions yet.")
        for t in self.transactions:
            print(f"  {t}")
        print(f"  Current Balance: Rs.{self.__balance}")
        print("-" * 40)
    
    def __str__(self):
        return f"Account({self.owner}, Balance: Rs.{self.__balance})"
```

```python
# Testing the BankAccount class
acc1 = BankAccount("Alice", 5000)
acc2 = BankAccount("Bob", 10000)

print()
# Performing transactions
acc1.deposit(3000)
acc1.withdraw(1500)
acc1.withdraw(10000)  # Should fail — insufficient funds
acc1.deposit(-500)    # Should fail — negative amount

acc2.deposit(5000)
acc2.withdraw(2000)

# Show transaction history
acc1.show_transactions()
acc2.show_transactions()

# Using __str__
print(acc1)
print(acc2)
```

```python
# SavingsAccount — inherits from BankAccount
class SavingsAccount(BankAccount):
    """A savings account with interest calculation."""
    
    def __init__(self, owner, balance=0, interest_rate=0.05):
        super().__init__(owner, balance)
        self.interest_rate = interest_rate  # e.g., 0.05 = 5%
    
    def apply_interest(self):
        """Calculate and add interest to balance."""
        interest = self.get_balance() * self.interest_rate
        self.deposit(interest)
        self.transactions.append(f"Interest Applied: +Rs.{interest:.2f} ({self.interest_rate*100}%)")
        return interest
    
    def __str__(self):
        return f"SavingsAccount({self.owner}, Balance: Rs.{self.get_balance()}, Rate: {self.interest_rate*100}%)"


# Testing SavingsAccount
print("=" * 50)
savings = SavingsAccount("Charlie", 100000, 0.08)  # 8% interest

savings.deposit(50000)
savings.withdraw(20000)

# Apply interest
interest = savings.apply_interest()
print(f"\nInterest earned: Rs.{interest:.2f}")

savings.show_transactions()
print(savings)
```
