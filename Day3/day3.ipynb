#### **Numbers in Python: int, float, and complex**

Python provides three main types of numeric data: **int**, **float**, and **complex**. These types allow you to perform a wide range of mathematical and scientific calculations.


#### **1. Integer (`int`)**
- **Definition**: Represents whole numbers without a fractional component.
- **Examples**: `1`, `0`, `-100`, `42`

#### **2. Floating-Point Number (`float`)**
- **Definition**: Represents numbers with a fractional (decimal) component.
- **Examples**: `1.23`, `0.0`, `-45.67`, `3.14159`

#### **3. Complex Number (`complex`)**
- **Definition**: Represents numbers in the form \(a + bj\), where:
  - \(a\) is the real part.
  - \(b\) is the imaginary part.
- **Examples**: `2+3j`, `-1+0j`, `3-4j`

````python
x=10
y=2.5
z=3+4j
````


#### **List Slicing**

List slicing allows you to extract portions of a list using a specific syntax.

#### Basic Syntax
```python
list[start:stop:step]
```

- **`start`**: The index where the slice begins (inclusive). Default is `0`.
- **`stop`**: The index where the slice ends (exclusive).
- **`step`**: The step size or interval between elements. Default is `1`.


````python
numbers = [0, 1, 2, 3, 4, 5]
subset = numbers[1:4]
print(subset)  
````

````python
print(numbers[:3])  # Omitting the start index
print(numbers[3:])  # Omitting the end index
print(numbers[:])   # Omitting both indexes
print(numbers[-4:-1]) # Negative Indexing
print(numbers[::-1])  # using Negative step (reverses the list)
````

#### **Dictionaries in Python**

A **dictionary** in Python is an unordered, mutable, and indexed collection of key-value pairs. Each key in a dictionary must be unique, and it is used to access its corresponding value.


#### **Creating a Dictionary**

#### **Syntax**:
```python
dictionary_name = {key1: value1, key2: value2, ...}

````python
person_1 = {"name": "Alice", "age": 25, "city": "New York"}
person_2 = dict(name="Bob", age=30, city="Los Angeles")
print(person_1)
````

### **Accessing Dictionary Values**

````python
print(person_1["name"])   # Accessing the value by key
print(person_1.get("age"))  # Accessing the value by key using get method
````

#### **What Is Difference ?**
The `get()` method retrieves a value for a key but avoids errors if the key doesn't exist by returning `None` or a default value.

````python
print(person_1["gender"])
````

````python
print(person_1.get("gender", "N/A"))  
````

````python
person = {"name": "Alice", "age": 25, "city": "New York"}
person["gender"] = "Female"  # Adding a new key-value pair
person["age"] = 26  # Updating the value of an existing key
age = person.pop("age")  # Removing a key-value pair
last_item = person.popitem()  # Removing the last key-value pair
del person["city"]  # Removing a key-value pair
person.clear()  # Removing all key-value pairs
````

````python
person = {"name": "Alice", "age": 25, "city": "New York"}
print(person.keys()) 
print(person.values()) 
print(person.items())
````

#### **Sets in Python**

A **set** in Python is an unordered collection of unique, immutable elements. Sets are used to store multiple items in a single variable, and they automatically remove duplicates.

#### **Key Features of Sets**
- **Unordered**: Elements are not stored in any specific order.
- **Unique**: Duplicate values are automatically removed.
- **Mutable**: You can add or remove items.
- **Immutable Elements**: Set elements must be immutable (e.g., strings, numbers, tuples).

#### **Creating a Set**
set_name = {item1, item2, item3, ...}

````python
my_set = {1, 2, 3, 4, 5}
empty_set = set()
char_set = set("hello")  # Output: {'h', 'e', 'l', 'o'}
````

````python
my_set = {1, 2, 3}
my_set.add(4)          # The `add()` method to adds a single element.
my_set.update([5, 6])  # The `update()` method to add multiple elements.
my_set.remove(6)       # The `remove()` method to remove an element (raises an error if the item doesn’t exist).
my_set.discard(3)      # The `discard()` method to remove an element without raising an error if it doesn’t exist.
item = my_set.pop()    # The `pop()` method to remove and return an arbitrary element.
my_set.clear()         # The `clear()` method to remove all elements.
````

### **Set Operations**

Python sets support standard set theory operations:

| Operation       | Description                                            | Syntax / Method                |
|------------------|--------------------------------------------------------|--------------------------------|
| Union           | Combines elements from both sets (no duplicates)       | `set1 | set2` or `set1.union(set2)` |
| Intersection    | Elements common to both sets                           | `set1 & set2` or `set1.intersection(set2)` |
| Difference      | Elements in `set1` but not in `set2`                   | `set1 - set2` or `set1.difference(set2)` |
| Symmetric Diff. | Elements in either set, but not in both                | `set1 ^ set2` or `set1.symmetric_difference(set2)` |
| Subset          | Checks if `set1` is a subset of `set2`                 | `set1 <= set2` or `set1.issubset(set2)` |
| Superset        | Checks if `set1` is a superset of `set2`               | `set1 >= set2` or `set1.issuperset(set2)` |
| Disjoint        | Checks if sets have no common elements                 | `set1.isdisjoint(set2)` |

````python
set1 = {1, 2, 3}
set2 = {3, 4, 5}
print(set1.union(set2))  # Output: {1, 2, 3, 4, 5}
print(set1.intersection(set2))  # Output: {3}
print(set1.difference(set2))  # Output: {1, 2}
print(set1.symmetric_difference(set2))  # Output: {1, 2, 4, 5}
````

#### **Tuples in Python**

A **tuple** in Python is an immutable, ordered collection of items. Once created, the elements of a tuple cannot be changed, added, or removed. Tuples are used to group related data and ensure that the values remain constant.

#### **Creating a Tuple**

```python
tuple_name = (item1, item2, item3, ...)
```

````python
empty_tuple = ()
my_tuple = (1, 2, 3)
mixed_tuple = (1, "Hello", 3.14)
````

#### **Accessing Tuple Elements**

````python
my_tuple = (10, 20, 30, 40, 50)
print(my_tuple[1])  # Indexing
print(my_tuple[-1]) # Negative Indexing
print(my_tuple[1:4]) # Slicing
````
