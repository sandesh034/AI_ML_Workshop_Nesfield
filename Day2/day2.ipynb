#### **1. The `if` Statement**

The `if` statement executes a block of code only if a specified condition is `True`.

````python
x=3

if(x>5):
    print("x is greater than 5")
````


#### **2. The `if-else` Statement**

The `if-else` statement provides an alternative block of code to execute if the condition is `False`.

#### **Syntax**:
```python
if condition:
    # code to execute if the condition is True
else:
    # code to execute if the condition is False
```

````python
x = 10
if x > 5:
    print("x is greater than 5")
else:
    print("x is not greater than 5")
````

#### **3. The `if-elif-else` Statement**

The `if-elif-else` structure allows you to check multiple conditions. Once a condition evaluates to `True`, the corresponding block is executed, and the remaining conditions are skipped.

#### **Syntax**:
```python
if condition1:
    # code to execute if condition1 is True
elif condition2:
    # code to execute if condition2 is True
else:
    # code to execute if all conditions are False
```


````python
x = 15
if x < 10:
    print("x is less than 10")
elif x == 15:
    print("x is equal to 15")
else:
    print("x is greater than 10 but not 15")
````

#### **4. Nested `if` Statements**

You can use `if` statements inside other `if` statements to test more complex conditions.

````python
x = 20
if x > 10:
    if x < 30:
        print("x is between 10 and 30")
````

#### **Ternary Operator in Python**

The **ternary operator** in Python is a concise way to write an `if-else` statement. It's often used to assign a value based on a condition in a single line of code.



### **Syntax**

```python
value_if_true if condition else value_if_false
```

- **`condition`**: The expression to evaluate.
- **`value_if_true`**: The value to return if the condition is `True`.
- **`value_if_false`**: The value to return if the condition is `False`.


##### **Key Points To Remember**
1. Simplifies short conditional assignments.
2. Useful for simple conditions where `if-else` would take up more lines.
3. For complex conditions or actions, using a ternary operator can reduce readability. In such cases, use regular `if-else` statements for clarity.
4. Keep the expressions short and simple.
5. Avoid chaining too many ternary operators; prefer `if-elif-else` for complex logic.



````python
x = 10
result = "Positive" if x > 0 else "Negative"
print(result)
````

#### **Loop in Python**

The `while` loop in Python is used to repeatedly execute a block of code as long as a given condition is `True`. Once the condition becomes `False`, the loop stops executing.



#### **Syntax**
```python
while condition:
    # Code to execute
```
- The **`condition`** is an expression that is evaluated before each iteration.
- If the **`condition`** is `True`, the loop continues to run.
- If the **`condition`** is `False`, the loop terminates.

#### **Key Points to Remember**
1. The condition is evaluated before each iteration. If the condition is `False` from the start, the code inside the `while` loop will not run.
2. If the condition never becomes `False`, the loop will run indefinitely. Always ensure the condition will eventually become `False`, 

````python
# While loop
count = 0
while count < 5:
    print(count)
    count += 1
````

````python
# Infinte While Loop
while True:
    print("This will print forever!")
````

#### **For Loops in Python**

A `for` loop in Python is used to iterate over a sequence (like a list, tuple, string, or range) and execute a block of code for each element in the sequence.


#### **Syntax**
```python
for variable in sequence:
    # code to execute for each element
```

- **`variable`**: Represents the current element in the sequence.
- **`sequence`**: The iterable object (e.g., list, string, range).





#### **Using `range()` Function**
- The `range()` function generates a sequence of numbers.
- It can take up to three arguments:  **`range(start, stop, step)`**

    - start: The starting value of the sequence (inclusive). Defaults to 0 if not specified.  (optional)
    - stop : The ending value of the sequence (exclusive). (required)
    - step : The increment (or decrement) between each number in the sequence. Defaults to 1. (optional)

````python
for x in range(5):
    print(x)
````

````python
for x in range(1,5):
    print(x)
````

````python
for x in range(1,5,2):
    print(x)
````

````python
# Iterate over a list

fruits = ["Apple", "Banana", "Cherry"]
for x in fruits:
    print(x)
````

````python
# Iterate over a string

for x in "ECAST":
    print(x)
````

#### **Break And Continue Statements**
- The `break` statement terminates the loop prematurely.

- The `continue` statement skips the current iteration and moves to the next.

````python
# Using break statement
for i in range(10):
    if i == 2:
        break
    print(i)
````

````python
# Using continue statement
for i in range(5):
    if i == 2:
        continue
    print(i)
````


#### **Using `else` in loop**
The `else` block is executed after the loop finishes, but it will not execute if the loop is terminated by a `break` statement.

````python
for i in range(5):
    print(i)
else:
    print("Loop completed")
````

* ALL the  `Break`, `Continue` and `Else` Can be used in while loop also. 

#### **Functions in Python**

A function in Python is a block of reusable code that performs a specific task. Functions help organize code, making it more modular, readable, and maintainable. Functions can accept parameters (inputs) and return a value (output).


#### **Defining a Function**

In Python, a function is defined using the `def` keyword, followed by the function name and parentheses. Any parameters are passed inside the parentheses, and the code block inside the function is indented.

#### **Syntax**
```python
def function_name(parameters):
    # Function body
    # Code to execute
    return result  
```

- **function_name**: Name of the function.
- **parameters**: Values or variables passed to the function. They are optional.
- **return**: The value returned by the function. It is also optional.

````python
def greet():
    print("Hello, World!")  #This function `greet()` does not take any parameters and simply prints "Hello, World!" when called.
````

````python
# Call the function
greet()
````

#### **Function with Parameters**

````python
#The function `greet()` takes a single parameter `name` and prints a personalized greeting.

def greet(name):   #parameter
    print(f"Hello, {name}!") 

greet("Thapathali")  #Argument
greet("ECAST")
````

#### **Function with Return Value**

````python
def add(a, b):
    sum=a+b
    return sum  # The function `add()` takes two parameters `a` and `b`, and returns their sum.

result = add(3, 4)
print(result)
````

#### **Function with Default Parameters**

We can provide default values for parameters. If no value is passed for those parameters, the default value is used.

````python
def greet(name="Guest"):
    print(f"Hello, {name}!")

greet("ECAST")  # Output: Hello, Alice!
greet()         # Output: Hello, Guest!
````

### **Function with Variable Number of Arguments**
Python allows you to pass a variable number of arguments to a function using `*args` (for non-keyword arguments) and `**kwargs` (for keyword arguments).

- The args accept the parameters in the form of tuple
- The kwargs accept the parameters in the form of dictionary

````python
def print_numbers(*args):
    print (args, type(args))
    sum=0
    for num in args:
        sum+=num
    print(sum)

def greet_person(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
````

````python
print_numbers(1, 2, 3, 4,5,6)
greet_person(name="Alice", age=30, city="New York")
````
