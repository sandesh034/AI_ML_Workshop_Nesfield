## 1. Series

A **Series** is a 1D labeled array — like a single column in a spreadsheet.

### Key Ways to Create a Series:
| Method | Description |
|--------|-------------|
| `pd.Series([...])` | From a Python list |
| `pd.Series({...})` | From a dictionary |
| `pd.Series(value, index=[...])` | Scalar with custom index |

```python
# Creating a Series from a list
s = pd.Series([10, 20, 30, 40, 50])
print("Series from list:")
print(s)
```

```python
# Series with custom index
s = pd.Series([85, 92, 78, 95], index=["Math", "Science", "English", "Computer"])
print("Series with custom index:")
print(s)
```

```python
# Series from a dictionary
population = pd.Series({
    "Kathmandu": 1442271,
    "Pokhara": 518452,
    "Lalitpur": 299198,
    "Biratnagar": 242548
})
print("Population Series:")
print(population)
```

```python
# Series properties
print(f"Values: {s.values}")
print(f"Index: {s.index.tolist()}")
print(f"Dtype: {s.dtype}")
print(f"Shape: {s.shape}")
print(f"Size: {s.size}")
```

```python
# Accessing elements
print(f"Math score: {s['Math']}")
print(f"First element: {s.iloc[0]}")
print(f"\nScores above 80:")
print(s[s > 80])
```

```python
# Series operations (vectorized)
print(f"Mean: {s.mean()}")
print(f"Max: {s.max()}")
print(f"Sum: {s.sum()}")
print(f"\nCurved scores (+5):")
print(s + 5)
```

## 2. DataFrames

A **DataFrame** is a 2D labeled data structure — like a spreadsheet or SQL table.

### Key Ways to Create a DataFrame:
| Method | Description |
|--------|-------------|
| `pd.DataFrame({...})` | From a dictionary of lists |
| `pd.DataFrame(list of dicts)` | From a list of dictionaries |
| `pd.DataFrame(numpy_array)` | From a NumPy array |
| `pd.read_csv("file.csv")` | From a CSV file |

```python
# Creating a DataFrame from a dictionary
data = {
    "Name": ["Aarav", "Sita", "Bikash", "Priya", "Rajan"],
    "Age": [21, 22, 20, 23, 21],
    "City": ["Kathmandu", "Pokhara", "Lalitpur", "Kathmandu", "Biratnagar"],
    "Score": [85, 92, 78, 95, 67]
}

df = pd.DataFrame(data)
print(df)
```

```python
# DataFrame from a list of dictionaries
records = [
    {"Name": "Aarav", "Age": 21, "Score": 85},
    {"Name": "Sita", "Age": 22, "Score": 92},
    {"Name": "Bikash", "Age": 20, "Score": 78},
]
df2 = pd.DataFrame(records)
print(df2)
```

```python
# DataFrame properties
print(f"Shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
print(f"Index: {df.index.tolist()}")
print(f"\nData types:")
print(df.dtypes)
```

```python
# head, tail, info, describe
print("First 3 rows:")
print(df.head(3))

print("\nLast 2 rows:")
print(df.tail(2))
```

```python
# info() — summary of the DataFrame
df.info()
```

```python
# describe() — statistical summary
df.describe()
```

## 4. Selecting Data

### Selecting Columns
- **Single column**: `df["column"]` → returns a Series
- **Multiple columns**: `df[["col1", "col2"]]` → returns a DataFrame

### Selecting Rows
- **`loc`**: Select by label (name)
- **`iloc`**: Select by integer position

```python
# Select a single column (returns Series)
print("Names:")
print(df["Name"])
```

```python
# Select multiple columns (returns DataFrame)
print("Name and Score:")
print(df[["Name", "Score"]])
```

```python
# loc — select by label
print("Row 0:")
print(df.loc[0])

print("\nRows 0-2, Name and Score columns:")
print(df.loc[0:2, ["Name", "Score"]])
```

```python
# iloc — select by integer position
print("First row:")
print(df.iloc[0])

print("\nFirst 3 rows, first 2 columns:")
print(df.iloc[:3, :2])
```

```python
# loc vs iloc: loc includes the endpoint, iloc does not
print("loc[0:2] — includes row 2:")
print(df.loc[0:2, "Name"])

print("\niloc[0:2] — excludes row 2:")
print(df.iloc[0:2, 0])
```

## 5. Filtering Data

Apply **conditions** to get only the rows you want — works like NumPy boolean indexing.

```python
# Filter: students with Score > 85
high_scorers = df[df["Score"] > 85]
print("Students with Score > 85:")
print(high_scorers)
```

```python
# Filter: students from Kathmandu
ktm_students = df[df["City"] == "Kathmandu"]
print("Students from Kathmandu:")
print(ktm_students)
```

```python
# Multiple conditions with & (AND)
result = df[(df["Score"] > 80) & (df["City"] == "Kathmandu")]
print("Score > 80 AND from Kathmandu:")
print(result)
```

```python
# Multiple conditions with | (OR)
result = df[(df["City"] == "Pokhara") | (df["City"] == "Lalitpur")]
print("From Pokhara OR Lalitpur:")
print(result)
```

```python
# Using isin() for multiple values
result = df[df["City"].isin(["Pokhara", "Lalitpur"])]
print("Using isin():")
print(result)
```

```python
# Using between()
result = df[df["Score"].between(75, 90)]
print("Score between 75 and 90:")
print(result)
```

```python
# String filtering
result = df[df["Name"].str.startswith("A")]
print("Names starting with 'A':")
print(result)
```

```python
# Using query() — a cleaner syntax
result = df.query("Score > 85 and City == 'Kathmandu'")
print("Using query():")
print(result)
```

## 7. Adding and Modifying Columns

```python
# Add a new column
df["Pass"] = df["Score"] >= 70
print("Added 'Pass' column:")
print(df.head())
```

```python
# Add a computed column
df["Curved_Score"] = df["Score"] + 5
print("Added 'Curved_Score' column:")
print(df[["Name", "Score", "Curved_Score"]].head())
```

```python
# Using apply() with a function
def assign_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    else:
        return "F"

df["Grade"] = df["Score"].apply(assign_grade)
print("Added 'Grade' column:")
print(df[["Name", "Score", "Grade"]])
```

```python
# Using np.where() for conditional columns
df["Category"] = np.where(df["Score"] >= 80, "High", "Low")
print("Added 'Category' column:")
print(df[["Name", "Score", "Category"]])
```

```python
# Rename columns
df_renamed = df.rename(columns={"Score": "Marks", "City": "Location"})
print("Renamed columns:")
print(df_renamed.columns.tolist())
```

```python
# Drop columns
df_clean = df.drop(columns=["Pass", "Curved_Score", "Category"])
print("After dropping columns:")
print(df_clean.head())
```

## 9. Handling Missing Values

Real-world data often has **missing values** (shown as `NaN` — Not a Number). Pandas provides tools to detect, remove, and fill them.

| Method | Description |
|--------|-------------|
| `df.isna()` / `df.isnull()` | Detect missing values (True/False) |
| `df.notna()` / `df.notnull()` | Detect non-missing values |
| `df.isna().sum()` | Count missing values per column |
| `df.dropna()` | Remove rows with missing values |
| `df.fillna(value)` | Replace missing values with a value |
| `df.interpolate()` | Fill missing values by interpolation |

```python
# Create a DataFrame with missing values
df_missing = pd.DataFrame({
    "Name": ["Aarav", "Sita", "Bikash", "Priya", "Rajan"],
    "Age": [21, np.nan, 20, 23, np.nan],
    "City": ["Kathmandu", "Pokhara", None, "Kathmandu", "Biratnagar"],
    "Score": [85, 92, np.nan, np.nan, 67]
})

print("DataFrame with missing values:")
print(df_missing)
```

```python
# Detect missing values
print("Missing values (True = missing):")
print(df_missing.isna())

print("\nCount of missing values per column:")
print(df_missing.isna().sum())

print(f"\nTotal missing values: {df_missing.isna().sum().sum()}")
```

```python
# Drop rows with any missing value
df_dropped = df_missing.dropna()
print("After dropna() — rows with ANY missing value removed:")
print(df_dropped)

print(f"\nOriginal shape: {df_missing.shape}")
print(f"After dropna shape: {df_dropped.shape}")
```

```python
# Drop rows where a specific column has missing values
df_dropped_score = df_missing.dropna(subset=["Score"])
print("After dropna(subset=['Score']) — only rows with missing Score removed:")
print(df_dropped_score)
```

```python
# Fill missing values with a specific value
df_filled = df_missing.fillna({"Age": df_missing["Age"].mean(), "City": "Unknown", "Score": 0})
print("After fillna() — missing values replaced:")
print(df_filled)
```

```python
# Forward fill and backward fill
df_ffill = df_missing["Score"].ffill()
print("Forward fill (use previous value):")
print(df_ffill)

df_bfill = df_missing["Score"].bfill()
print("\nBackward fill (use next value):")
print(df_bfill)
```

```python
# Interpolate missing values (linear interpolation)
df_interp = df_missing["Score"].interpolate()
print("Interpolated Score:")
print(df_interp)
```

## 10. Handling Unique Values & Duplicates

Pandas provides tools to find **unique values**, count them, and handle **duplicate rows**.

| Method | Description |
|--------|-------------|
| `df["col"].unique()` | Array of unique values |
| `df["col"].nunique()` | Count of unique values |
| `df["col"].value_counts()` | Frequency of each value |
| `df.duplicated()` | Boolean mask of duplicate rows |
| `df.drop_duplicates()` | Remove duplicate rows |
| `df["col"].replace()` | Replace specific values |

```python
# Unique values in a column
print("Unique cities:", df["City"].unique())
print(f"Number of unique cities: {df['City'].nunique()}")
print(f"\nUnique subjects: {df['Subject'].unique()}")
print(f"Number of unique subjects: {df['Subject'].nunique()}")
```

```python
# Value counts — frequency of each value (sorted by count)
print("City value counts:")
print(df["City"].value_counts())

print("\nSubject value counts:")
print(df["Subject"].value_counts())

# Normalized value counts (proportions)
print("\nCity proportions:")
print(df["City"].value_counts(normalize=True))
```

```python
# Detecting and removing duplicates
df_with_dupes = pd.DataFrame({
    "Name": ["Aarav", "Sita", "Aarav", "Bikash", "Sita"],
    "Age": [21, 22, 21, 20, 22],
    "Score": [85, 92, 85, 78, 92]
})

print("DataFrame with duplicates:")
print(df_with_dupes)

print("\nDuplicate rows (True = duplicate):")
print(df_with_dupes.duplicated())

print(f"\nNumber of duplicates: {df_with_dupes.duplicated().sum()}")
```

```python
# Remove duplicates
df_no_dupes = df_with_dupes.drop_duplicates()
print("After drop_duplicates():")
print(df_no_dupes)

# Remove duplicates based on a specific column
df_unique_names = df_with_dupes.drop_duplicates(subset=["Name"], keep="first")
print("\nKeep first occurrence of each Name:")
print(df_unique_names)
```

```python
# Replace specific values
df_replaced = df.copy()
df_replaced["City"] = df_replaced["City"].replace({"Biratnagar": "Morang", "Lalitpur": "Patan"})
print("After replacing city names:")
print(df_replaced["City"].value_counts())

# Map values using a dictionary
grade_map = {"Math": "M", "Science": "S"}
df_replaced["Subject_Code"] = df_replaced["Subject"].map(grade_map)
print("\nMapped Subject to Subject_Code:")
print(df_replaced[["Name", "Subject", "Subject_Code"]].head())
```

The Iris dataset is a classic ML dataset with 150 samples of iris flowers.

```python
# Exercise 3: Find flowers with sepal_length > 7
large_sepals = iris[iris["sepal_length"] > 7.0]
print(f"Flowers with sepal_length > 7: {len(large_sepals)}")
print(large_sepals)
```

```python
# Exercise 4: Sort by petal_length (descending) and show top 10
print("Top 10 by petal_length:")
print(iris.nlargest(10, "petal_length")[["species", "petal_length", "petal_width"]])
```
